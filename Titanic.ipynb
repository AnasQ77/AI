# How many people died or went missing during the incident of the Titanic? 



#### I will conduct a consensus to measure how many people there were before the incident and after the incident occured to compare how many survivors there were, this will help me find the answer to how many people went missing or died.

```python
import pandas as pd

url = "https://drive.google.com/uc?export=download&id=1ViNYIL-Cgj3rmuNtyVVc0SXcoIbFkF81"

df = pd.read_csv(url)

df.head()
```

```python
titanic = df[df["Description"].str.contains("Titanic", case=False, na=False)]

titanic
```

```python
titanic["Fatalities"]
```

```python
total_fatalities = titanic["Fatalities"].sum()

print("Number of people who died or went missing:", total_fatalities)
```
